```python
def add_args_to_dict(dict, **kwargs):
    for key, value in kwargs.items():
        dict[key] = value
```

```python
def check_Unique(layername):  # Dont add a layer twice
    global HomaTect_temp_list
    global HomaTect_layer_names
    global HomaTect_layers_list
    if layername not in HomaTect_layer_names: 
        HomaTect_layer_names.append(layername)
        ########################### print("I'm addeing a unique layer...", layername)
        HomaTect_temp_list.append(HomaTect_layer_args_dict.copy())
```

```python
@keras_export(["keras.Sequential", "keras.models.Sequential"])
class new_Sequential(keras.models.Sequential):
    global HomaTect_layers_list
    def __init__(self, layers=None, trainable=True, name=None):
        HomaTect_layers_list.clear() # Make sure it doesnt contain any layers from previous runs
        ###################################### print("In sequential and there are already ", len(HomaTect_layers_list), "layers")
        self._functional = None
        self._layers = []
        super().__init__(trainable=trainable, name=name)
        if layers:
            ################################################# print("there are layers: ", layers)
            ################################################# print("there are layers in dict: ", HomaTect_temp_list)
            for layer in HomaTect_temp_list:
                ############################################ print("Adding because it's in layers...")
                self.add(layer, rebuild=False)
            HomaTect_temp_list.clear()
            self._maybe_rebuild()
        

    def add(self, layer, rebuild=True):
        global HomaTect_layers_list
        global HomaTect_layer_args_dict
        ######################################################### print("In Add....")
        #------------------------------------ if isinstance(layer, dict) and layer["layer"] != "InputLayer": # Input layer is input
        if isinstance(layer, dict) and bool(layer): # Dont add empty dict in case of InputLayer
            HomaTect_layers_list.append(layer.copy())
            # --------------------print("Added using sequential arguments ", layer)
        #------------------------------------ elif isinstance(layer, Layer) and HomaTect_layer_args_dict["layer"] != "InputLayer":
        elif isinstance(layer, Layer) and bool(HomaTect_layer_args_dict): # Dont add empty dict in case of InputLayer
            # Add all layers including Activation to preserve layer numbers
            ########################################### print("Added using add method", HomaTect_layer_args_dict)
            HomaTect_layers_list.append(HomaTect_layer_args_dict.copy())
        HomaTect_layer_args_dict.clear()
        # ----------------------- print("%%%%%%%%%%%%%%%%%%%% list is: ", HomaTect_layers_list) # print whether it's activation or any other layer

        self._layers.append(layer)
        if rebuild:
            self._maybe_rebuild()
        else:
            self.built = False
            self._functional = None


keras.models.Sequential = keras.Sequential = Sequential = new_Sequential
tf.keras.models.Sequential = tf.keras.Sequential = Sequential = new_Sequential
```

```python
@traceback_utils.filter_traceback
@tracking.no_automatic_dependency_tracking
def debugger_compile(
        self,
        optimizer="rmsprop",
        loss=None,
        loss_weights=None,
        metrics=None,
        weighted_metrics=None,
        run_eagerly=False,
        steps_per_execution=1,
        jit_compile="auto",
        auto_scale_loss=True,
    ):
    if callable(loss): # If loss argument is a function
        if loss.__qualname__ in ["categorical_crossentropy", "binary_crossentropy", "sparse_categorical_crossentropy", "mean_squared_error"]:
            loss = loss.__qualname__

    add_args_to_dict(
        HomaTect_compile_args_dict,
        optimizer=optimizer,
        loss=loss,
        loss_weights=loss_weights,
        metrics=metrics,
        weighted_metrics=weighted_metrics,
        run_eagerly=run_eagerly,
        steps_per_execution=steps_per_execution,
        jit_compile=jit_compile,
        auto_scale_loss=auto_scale_loss,
    )
   
    ################################################ print("compile args are: ", HomaTect_compile_args_dict)


Sequential.compile = debugger_compile
```

```python
def Null_Bias_Initialization():
    for i, layer in enumerate(HomaTect_layers_list):
        if "bias_initializer" not in layer:
            continue
        bias_init = layer["bias_initializer"]
        violation = False

        # String initializers
        if isinstance(bias_init, str):
            if bias_init.lower() != "zeros":
                violation = True

        # Initializer objects
        elif hasattr(bias_init, "__class__"):
            initializer_name = bias_init.__class__.__name__.lower()
            if initializer_name == "zeros":
                violation = False
            elif initializer_name == "constant":
                try: # Keras Constant initializer
                    value = bias_init.value
                    if hasattr(value, "numpy"): # Handle tf.Tensor, numpy scalar, etc.
                        value = value.numpy()
                    violation = value != 0
                except:
                    violation = True
            else:
                violation = True
        if violation:
            if isinstance(bias_init, str):
                initializer_name = bias_init
            else:
                initializer_name = bias_init.__class__.__name__
            print(f"Warning: Layer {i+1}: Biases are initialized using '{initializer_name}', which produces non-zero initial bias values. It is suggested to use the 'Zeros' initializer for biases.")
```

```python
def Hidden_layer_Activation_initializer():
    global HomaTect_layers_list
    global HomaTect_output
    activation_defined = None
    outputlayer = True
    actid = None
    for layerid, layer in reversed(list(enumerate(HomaTect_layers_list))):
        if (layer["layer"] in ["Dense", "Conv1D", "Conv2D", "Conv3D"]):
            if (layer["kernel_initializer"] in ["zeros", "ones", "Zeros", "Ones"]) or (layer["kernel_initializer"].__class__.__name__.lower() in ["zeros", "ones", "constant"]):
                print(f"Layer{layerid+1}: It's suggested not to use constant kernel initializer")                
            
            if outputlayer == True: # keep output layer layerid
                HomaTect_output = layerid
            else: # dont do the check for output layer, only hidden layers
                if  layer["activation"] in [None, "linear"] and activation_defined in [None, "linear"]:
                    if activation_defined == "linear":
                        print(f"Layer{actid + 1}: layer needs a non linear activation function. you can use relu.")
                    else: # activation_define == none
                        print(f"Layer{layerid + 1}: layer needs a non linear activation function.you can use relu.")
                if  (layer["activation"] not in [None, "linear"]) and (activation_defined not in [None, "linear"]):
                    if layer["activation"] != activation_defined:
                        print(f"Layer{layerid+1}: Activation function is redundant and prevents the activation function in layer{actid} from producing its full output range.")
                if  layer["activation"] in ["sigmoid", "softmax"]:
                    print(f"Layer{layerid + 1}: {layer['activation']} is not an appropriate activation function for hidden layers, use relu instead.")
                if activation_defined in ["sigmoid", "softmax"]:
                    print(f"Layer{actid+1}: {layer['activation']} is not an appropriate activation function for hidden layers, use relu instead.")
                
            activation_defined = None
            outputlayer = False
                
        if layer["layer"] == "Activation":
            activation_defined = layer["activation"]
            actid = layerid
```

```python
def Dropout_rate_after_Maxpooling_BatchNorm():
    global HomaTect_layers_list
    dropout = True
    for layerid in range (0, len(HomaTect_layers_list) - 1):
        if HomaTect_layers_list[layerid]["layer"] == "Dropout":
            if HomaTect_layers_list[layerid]["rate"] > 0.5:
                print(f"Warning: Layer {layerid + 1}: Dropout rate is suggested not to be > 0.5")
                
        if HomaTect_layers_list[layerid]["layer"] == "Dropout":
            for j in range(layerid + 1, len(HomaTect_layers_list)):
                next_layer = HomaTect_layers_list[j]["layer"]
        
                # Skip activation layers
                if next_layer in ["Activation", "ReLU", "LeakyReLU", "PReLU", "ELU", "Softmax"]:
                    continue
        
                # Found MaxPooling
                if next_layer in ["MaxPooling1D", "MaxPooling2D", "MaxPooling3D"]:
                    print(f"Warning: Layer {j + 1}: It's suggested to add MaxPooling layer before Dropout layer, not after")
                    break
        
                # Stop search when reaching a new structural stage
                if next_layer in [
                    "Dense",
                    "Conv1D", "Conv2D", "Conv3D",
                    "Flatten",
                    "AveragePooling1D", "AveragePooling2D", "AveragePooling3D",
                    "GlobalAveragePooling1D", "GlobalAveragePooling2D", "GlobalAveragePooling3D",
                    "GlobalMaxPooling1D", "GlobalMaxPooling2D", "GlobalMaxPooling3D"]:
                    break  
        
        if HomaTect_layers_list[layerid]["layer"] == "Dropout":
            for j in range(layerid + 1, len(HomaTect_layers_list)):
                next_layer = HomaTect_layers_list[j]["layer"]
                if next_layer in [ # Skip activation layers
                    "Activation", "ReLU", "LeakyReLU", "PReLU", "ELU", "Softmax"]:
                    continue
                if next_layer == "BatchNormalization":
                    print(f"Warning: Layer {j + 1}: It's suggested to add BatchNormalization layer before Dropout layer, not after")
                    break
            
                # Stop when we reach a new structural stage
                if next_layer in [
                    "Dense",
                    "Conv1D", "Conv2D", "Conv3D",
                    "Flatten",
                    "MaxPooling1D", "MaxPooling2D", "MaxPooling3D",
                    "AveragePooling1D", "AveragePooling2D", "AveragePooling3D",
                    "GlobalAveragePooling1D", "GlobalAveragePooling2D", "GlobalAveragePooling3D",
                    "GlobalMaxPooling1D", "GlobalMaxPooling2D", "GlobalMaxPooling3D"]:
                    break
```

```python
def No_Bias_after_BatchNormalization():
    global HomaTect_layers_list
    activation_layers = ["Activation", "ReLU", "LeakyReLU", "PReLU", "ELU", "Softmax"]
    learning_layers = ["Dense", "Conv1D", "Conv2D", "Conv3D"]
    for layerid in range(len(HomaTect_layers_list)):
        current_layer = HomaTect_layers_list[layerid]
        if current_layer["layer"] not in learning_layers:
            continue

        if not current_layer.get("use_bias", True):
            continue

        for j in range(layerid + 1, len(HomaTect_layers_list)):
            next_layer = HomaTect_layers_list[j]["layer"]
            if next_layer in activation_layers:
                continue

            if next_layer == "BatchNormalization": 
                if HomaTect_layers_list[j].get("center", True): # BatchNorm must actually learn a shift
                    print(f"Warning: Layer {layerid + 1}: It's suggested to remove the unnecessary bias from this layer because it is followed by a BatchNormalization layer.")
                break

            # Stop when a new learning stage starts
            if next_layer in learning_layers + [
                "Flatten",
                "MaxPooling1D", "MaxPooling2D", "MaxPooling3D", "AveragePooling1D", "AveragePooling2D", "AveragePooling3D",
                "GlobalAveragePooling1D", "GlobalAveragePooling2D", "GlobalAveragePooling3D",
                "GlobalMaxPooling1D", "GlobalMaxPooling2D", "GlobalMaxPooling3D"]:
                break
```

```python
def Max_Info_Utilization():
    global HomaTect_layers_list
    poolingcount = 0
    convcount = 0
    if len(HomaTect_layers_list) >= 10:
        for layerid in range (0, len(HomaTect_layers_list)):
            if HomaTect_layers_list[layerid]["layer"] in ["MaxPooling1D", "MaxPooling2D", "MaxPooling3D", "AveragePooling1D", "AveragePooling2D", "AveragePooling3D", "GlobalAveragePooling1D", "GlobalAveragePooling2D", "GlobalAveragePooling3D", "GlobalMaxPooling1D", "GlobalMaxPooling2D", "GlobalMaxPooling3D"]:
                poolingcount += 1
            if HomaTect_layers_list[layerid]["layer"] in ["Conv1D", "Conv2D", "Conv3D"]:
                convcount += 1
        if convcount + poolingcount < 3 * poolingcount:
            print(f"Warning: Excessive pooling detected. The model contains {convcount} convolution layer(s) and {poolingcount} pooling layer(s). Pooling layers account for more than one third of all convolution/pooling layers. Excessive down-sampling may discard useful feature information too early. Consider reducing the number of pooling layers or adding more convolution layers between pooling operations.")
```

```python
def Max_Pooling_Domination():
    global HomaTect_layers_list
    for layerid in range(len(HomaTect_layers_list)):
        if HomaTect_layers_list[layerid]["layer"] in ["AveragePooling1D", "AveragePooling2D", "AveragePooling3D"]:
            print(f"Warning: Layer {layerid+1}: Max-pooling is a commonly used down-sampling strategy. It's suggested to replace AveragePooling layer with MaxPooling layer")

        if HomaTect_layers_list[layerid]["layer"] in ["Conv1D", "Conv2D", "Conv3D"]:
            strides = HomaTect_layers_list[layerid]["strides"]
            if isinstance(strides, int):
                strided = strides > 1
            else:
                strided = any(s > 1 for s in strides)
            if strided:
                print(f"Warning: Layer {layerid+1}: Max-pooling is a commonly used down-sampling strategy. It's suggested to replace convolution strides with a MaxPooling layer. Currently strides is {strides}")
```

```python
def FeaturemapsNum_LocalCorrelation_Homogeneity():
    global HomaTect_layers_list
    featuremaps = 0
    prev_kernel_area_1d = 0
    prev_kernel_area_2d = 0
    prev_kernel_area_3d = 0
    for layerid in range (0, len(HomaTect_layers_list)):
        if HomaTect_layers_list[layerid]["layer"] in ["Conv1D", "Conv2D", "Conv3D"]:
            if featuremaps > HomaTect_layers_list[layerid]["filters"]:
                print(f'Warning: Layer{layerid+1}: It is suggested that the number of feature maps be increasing to compensate for the loss of representational expressiveness caused by the continuous decreasing of the spatial resolution of the learned feature maps.')
            featuremaps = HomaTect_layers_list[layerid]["filters"]

            if HomaTect_layers_list[layerid]["layer"] == "Conv1D":
                kernel_size = HomaTect_layers_list[layerid]["kernel_size"]
                
                if isinstance(kernel_size, (list, tuple)):
                    kernel_size = kernel_size[0]
                current_kernel_area = kernel_size
                    
                if kernel_size >= 5:
                    print(f"Warning: Layer{layerid+1}: kernel_size = {kernel_size}. It is suggested that the Deep CNN should favor blocks of 2, 3 or even 4 homogeneous convolutional layers with similar characteristics. larger kernels can be turned into a groupe of smaller ones to enhances the nonlinearity. For instance, one 5 × 5 can be replaced by two 3 × 3 or four 2 × 2 kernels.")
                
                if prev_kernel_area_1d > current_kernel_area:
                    print(f"Warning: Layer{layerid+1}: It is suggested that the local window size for spatial filtering be generally increasing or stay the same throughout the convolutional layers to preserve locality.")
                prev_kernel_area_1d = current_kernel_area     

            elif HomaTect_layers_list[layerid]["layer"] == "Conv2D":
                kernel_size = HomaTect_layers_list[layerid]["kernel_size"]
                if isinstance(kernel_size, int):
                    currentwindow = (kernel_size, kernel_size)
                else:
                    currentwindow = kernel_size
                current_kernel_area = (currentwindow[0] * currentwindow[1])

                if currentwindow[0] >= 5 or currentwindow[1] >= 5:
                    print(f"Warning: Layer {layerid+1}: kernel_size = {currentwindow}. It is suggested that the Deep CNN should favor blocks of 2, 3 or even 4 homogeneous convolutional layers with similar characteristics. Larger kernels can be turned into a group of smaller ones to enhance nonlinearity. For instance, one 5×5 can be replaced by two 3×3 or four 2×2 kernels.")

                if prev_kernel_area_2d > current_kernel_area:
                    print(f"Warning: Layer {layerid+1}: It is suggested that the local window size for spatial filtering be generally increasing or stay the same throughout the convolutional layers to preserve locality.")
                prev_kernel_area_2d = current_kernel_area

            elif HomaTect_layers_list[layerid]["layer"] == "Conv3D":
                kernel_size = HomaTect_layers_list[layerid]["kernel_size"]

                if isinstance(kernel_size, int):
                    currentwindow = (kernel_size, kernel_size, kernel_size)
                else:
                    currentwindow = kernel_size
                current_kernel_area = (currentwindow[0] * currentwindow[1] * currentwindow[2])

                if (currentwindow[0] >= 5 or currentwindow[1] >= 5 or currentwindow[2] >= 5):
                    print(f"Warning: Layer {layerid+1}: kernel_size = {currentwindow}. It is suggested that the Deep CNN should favor blocks of 2, 3 or even 4 homogeneous convolutional layers with similar characteristics. Larger kernels can be turned into a group of smaller ones to enhance nonlinearity. For instance, one 5×5 can be replaced by two 3×3 or four 2×2 kernels.")

                if prev_kernel_area_3d > current_kernel_area:
                    print(f"Warning: Layer {layerid+1}: It is suggested that the local window size for spatial filtering be generally increasing or stay the same throughout the convolutional layers to preserve locality.")
                prev_kernel_area_3d = current_kernel_area   
```

```python
def Compute_Output_shapes(user_input_shape):
    global HomaTect_input_shape
    global HomaTect_layers_list
    InputShape = HomaTect_input_shape
    dense_width = None
    feature_maps_area = None
    HomaTect_previous_layer_negative = False
    for layerid in range(0, len(HomaTect_layers_list)):
        OutputShape = Output_Shape(layerid, InputShape, user_input_shape)
        
        for element in InputShape:
            if element < 0:
                print(f"Warning: Layer{layerid+1} Negative dimensions.This is due to previous layer's (layer{layerid}) negative dimensions and resolving the root issue fixes this one as well.")
                HomaTect_previous_layer_negative = True
                break
        
        InputShape = OutputShape
        for dimension in OutputShape: # If more that one dimension is negative, print the error once
            if dimension <= 0 and HomaTect_previous_layer_negative == False:
                print(f"Layer{layerid+1}: Negative dimensions: A {HomaTect_layers_list[layerid]['layer']} layer requires input_shape >= kernel_size for all dimensions. For example try making your kernels smaller, using 'padding = same', ...")
                HomaTect_previous_layer_negative = True
                break
            # ----------------------- elif dimension <= 0:
                # ------------------ print(f"Warning: layer{layerid+1} Negative dimensions.This is probably due to previous layer's (layer{layerid}) negative dimensions and resolving the root issue may fix this one as well.")
                # ------------------ HomaTect_previous_layer_negative = True
                # ------------------ break
        if HomaTect_layers_list[layerid]["layer"] == "Dense":
            if dense_width == None:
                dense_width = HomaTect_layers_list[layerid]["units"]
            elif HomaTect_layers_list[layerid]["units"] > dense_width:
                print(f"Warning: Layer{layerid+1}: It's suggested that the width of fully-connected units be decreasing, while the model is using {HomaTect_layers_list[layerid]['units']} units after {dense_width} units in the previous layer")
                dense_width = HomaTect_layers_list[layerid]["units"]
        elif HomaTect_layers_list[layerid]["layer"] in ["Conv1D", "Conv2D", "Conv3D", "MaxPooling1D", "MaxPooling2D", "MaxPooling3D", "AveragePooling1D", "AveragePooling2D", "AveragePooling3D"]:
            temparea = 1
            for dimension in OutputShape:
                temparea = temparea * dimension
            if feature_maps_area == None:
                feature_maps_area = temparea
            elif temparea > feature_maps_area:
                print(f"Warning: Layer{layerid+1}: It's suggested that the The area of feature maps be decreasing.")
                feature_maps_area = temparea
```

```python
def Output_Shape(layerid, InputShape, user_input_shape):
    global HomaTect_layers_list
    global HomaTect_first_conv  
    if HomaTect_layers_list[layerid]["layer"] == "Dense":
        # Input shape: N-D tensor with shape: (batch_size, ..., input_dim).
        # Output shape: N-D tensor with shape: (batch_size, ..., units).
        neg = False
        if len(InputShape) > 1:
            feature_map_source = False
            previous_layer = layerid - 1
            while previous_layer >= 0:
                previous_type = HomaTect_layers_list[previous_layer]["layer"]
                if previous_type in ["Flatten", "Reshape"]: # rank greater that 2 was probably intentional
                    break
                if previous_type in ["Activation", "Dropout", "BatchNormalization", "elu", "leakyrelu", "prelu", "relu", "softmax"]:
                    previous_layer -= 1
                    continue
                if previous_type in ["Conv1D", "Conv2D", "Conv3D", "MaxPooling1D", "MaxPooling2D", "MaxPooling3D", "AveragePooling1D", "AveragePooling2D", "AveragePooling3D"]:
                    feature_map_source = True
                    break
                break
            if feature_map_source:
                print(f"Layer{layerid+1}: The Dense layer receives a multi-dimensional feature map without a preceding Flatten layer. If the Dense layer is intended to operate on the complete feature vector, a Flatten layer should be added before it.")

        for element in InputShape:
            if element < 0:
                neg = True
        if neg == True: # If input to dense is negative, it also returns negative
            OutputShape = InputShape[:-1] + (-HomaTect_layers_list[layerid]["units"],)
        else:
            OutputShape = InputShape[:-1] + (HomaTect_layers_list[layerid]["units"],)
        return OutputShape
        
    if HomaTect_layers_list[layerid]["layer"] in ["Conv1D", "Conv2D", "Conv3D"]:
        if HomaTect_layers_list[layerid]["layer"] == "Conv1D":
            # filters: int
            # kernel_size: int or tuple/list of 1 integer
            # strides: int or tuple/list of 1 integer
            # padding: "valid", "same" or "causal"(case-insensitive).
            # data_format: "channels_last" corresponds to inputs with shape (batch, steps, features) while
                         # "channels_first" corresponds to inputs with shape (batch, features, steps). If you never set it, then it will be "channels_last".
            # Input shape: If data_format="channels_last": A 3D tensor with shape: (batch_shape, steps, channels)
                        #  If data_format="channels_first": A 3D tensor with shape: (batch_shape, channels, steps)
            # Output shape: If data_format="channels_last": A 3D tensor with shape: (batch_shape, new_steps, filters)
                        # If data_format="channels_first": A 3D tensor with shape: (batch_shape, filters, new_steps)        
            if len(user_input_shape) != 1 or len(InputShape) != 2: 
                raise ValueError(f" layer{layerid+1}: A Conv1D layer expects a 3D tensor, so 'user_input_shape' will be a 1D tensor and the input_shape will be a 2D tensor. Cant proceed with the debugging. Check the defined input shape")
            
            if isinstance(HomaTect_layers_list[layerid]["kernel_size"], (list, tuple)):
                kernel_size = HomaTect_layers_list[layerid]["kernel_size"][0]
            elif isinstance(HomaTect_layers_list[layerid]["kernel_size"], int):
                kernel_size = HomaTect_layers_list[layerid]["kernel_size"]
            if isinstance(HomaTect_layers_list[layerid]["strides"], (list, tuple)):
                strides = HomaTect_layers_list[layerid]["strides"][0]
            elif isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides = HomaTect_layers_list[layerid]["strides"]
                

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # ****** if len(InputShape) != 2: # Check input shape of the Conv1D layer withut batch_size
                    print(f"Layer{layerid+1}: A Conv1D layer with channels_first format, expects a 3D tensor with shape: (batch_size, channels, steps) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}.")
                    InputShape = InputShape[-1:] + InputShape[:-1] # Turn it into channels_first format
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temp = -(InputShape[1] // -strides ) # To make it work as ceiling
                    return (HomaTect_layers_list[layerid]["filters"], temp)
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temp = ((InputShape[1] - kernel_size) // strides) + 1                   
                    return (HomaTect_layers_list[layerid]["filters"], temp)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True: # check if channels_first/channels_last is applied correctly
                # ************* if len(InputShape) != 2: # Check input shape of the Conv1D layer without batch_size
                    print(f"Layer{layerid+1}: A Conv1D layer with channels_last format, expects a 3D tensor with shape: (batch_size, steps, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1] # Turn it into channels_last format
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same":
                    temp = -(InputShape[0] // -strides ) # To make it work as ceiling
                    return (temp, HomaTect_layers_list[layerid]["filters"])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temp = ((InputShape[0] - kernel_size) // strides) + 1
                    return (temp, HomaTect_layers_list[layerid]["filters"])

        if HomaTect_layers_list[layerid]["layer"] == "Conv2D":
            # filters: int
            # kernel_size:  int or tuple/list of 2 integers
            # strides: int or tuple/list of 2 integers
            # padding: "valid", "same".
            # data_format: "channels_last" corresponds to inputs with shape (batch_size, height, width, channels) while
                         # "channels_first" corresponds to inputs with shape (batch_size, channels, height, width). If you never set it, then it will be "channels_last".
            # Input shape: data_format = "channels_last": A 4D tensor with shape: (batch_size, height, width, channels)
                        # data_format = "channels_first": A 4D tensor with shape: (batch_size, channels, height, width)
            # Output shape: data_format="channels_last": A 4D tensor with shape: (batch_size, new_height, new_width, filters)
                        # data_format="channels_first": A 4D tensor with shape: (batch_size, filters, new_height, new_width)
            
            if len(user_input_shape) != 2 or len(InputShape) != 3: 
                raise ValueError(f" layer{layerid+1}: A Conv2D layer expects a 4D tensor, so 'user_input_shape' will be a 2D tensor and the input_shape will be a 3D tensor. Cant proceed with the debugging. Check the defined input shape")

            if isinstance(HomaTect_layers_list[layerid]["kernel_size"], (list, tuple)):
                kernel_height = HomaTect_layers_list[layerid]["kernel_size"][0]
                kernel_width = HomaTect_layers_list[layerid]["kernel_size"][1]
            elif isinstance(HomaTect_layers_list[layerid]["kernel_size"], int):
                kernel_height = HomaTect_layers_list[layerid]["kernel_size"]
                kernel_width = HomaTect_layers_list[layerid]["kernel_size"]
                
            if isinstance(HomaTect_layers_list[layerid]["strides"], (list, tuple)):
                strides_height = HomaTect_layers_list[layerid]["strides"][0]
                strides_width = HomaTect_layers_list[layerid]["strides"][1]
            if isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides_height = HomaTect_layers_list[layerid]["strides"]
                strides_width = HomaTect_layers_list[layerid]["strides"]

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # ***************** if len(InputShape) != 3: # Check input shape of the Conv2D layer without batch_size
                    print(f"Layer{layerid+1}: A Conv2D layer with channels_first format, expects a 4D tensor with shape: (batch_size, channels, height, width) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}")
                    InputShape = InputShape[-1:] + InputShape[:-1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temph = -(InputShape[1] // -strides_height) # To make it work as ceiling
                    tempw = -(InputShape[2] // -strides_width)
                    return (HomaTect_layers_list[layerid]["filters"], temph, tempw)
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = ((InputShape[1] - kernel_height) // strides_height) + 1
                    tempw = ((InputShape[2] - kernel_width) // strides_width) + 1
                    return (HomaTect_layers_list[layerid]["filters"], temph, tempw)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True:
                # ******************************* if len(InputShape) != 3: # Check input shape of the Conv2D layer
                    print(f"Layer{layerid+1}: A Conv2D layer with channels_last format, expects a 4D tensor with shape: (batch_shape, height, width, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same":
                    temph = -(InputShape[0] // -strides_height)
                    tempw = -(InputShape[1] // -strides_width)
                    return (temph, tempw, HomaTect_layers_list[layerid]["filters"])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = ((InputShape[0] - kernel_height) // strides_height) + 1
                    tempw = ((InputShape[1] - kernel_width) // strides_width) + 1
                    return (temph, tempw, HomaTect_layers_list[layerid]["filters"])

        
        if HomaTect_layers_list[layerid]["layer"] == "Conv3D":
            # filters: int
            # kernel_size:  int or tuple/list of 3 integers
            # strides: int or tuple/list of 3 integers
            # padding: "valid", "same".
            # data_format: "channels_last" corresponds to inputs with shape (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels) while
                         # "channels_first" corresponds to inputs with shape (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3). If you never set it, then it will be "channels_last".
            # Input shape: data_format = "channels_last": A 5D tensor with shape: (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels)
                        # data_format = "channels_first": A 5D tensor with shape: (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3)
            # Output shape: data_format="channels_last": A 5D tensor with shape: (batch_size, new_spatial_dim1, new_spatial_dim2, new_spatial_dim3, filters)
                        # data_format="channels_first": A 5D tensor with shape: (batch_size, filters, new_spatial_dim1, new_spatial_dim2, new_spatial_dim3)
            
            if len(user_input_shape) != 3 or len(InputShape) != 4: 
                raise ValueError(f" layer{layerid+1}: A Conv3D layer expects a 5D tensor, so 'user_input_shape' will be a 3D tensor and the input_shape will be a 4D tensor. Cant proceed with the debugging. Check the defined input shape")

            if isinstance(HomaTect_layers_list[layerid]["kernel_size"], (list, tuple)):
                kernel_height = HomaTect_layers_list[layerid]["kernel_size"][0]
                kernel_width = HomaTect_layers_list[layerid]["kernel_size"][1]
                kernel_depth = HomaTect_layers_list[layerid]["kernel_size"][2]
            elif isinstance(HomaTect_layers_list[layerid]["kernel_size"], int):
                kernel_height = HomaTect_layers_list[layerid]["kernel_size"]
                kernel_width = HomaTect_layers_list[layerid]["kernel_size"]
                kernel_depth = HomaTect_layers_list[layerid]["kernel_size"]
                
            if isinstance(HomaTect_layers_list[layerid]["strides"], (list, tuple)):
                strides_height = HomaTect_layers_list[layerid]["strides"][0]
                strides_width = HomaTect_layers_list[layerid]["strides"][1]
                strides_depth = HomaTect_layers_list[layerid]["strides"][2]
            if isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides_height = HomaTect_layers_list[layerid]["strides"]
                strides_width = HomaTect_layers_list[layerid]["strides"]
                strides_depth = HomaTect_layers_list[layerid]["strides"]

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # ******************* if len(InputShape) != 4: # Check input shape of the Conv3D layer without batch_size
                    print(f"Layer{layerid+1}: A Conv3D layer with channels_first format, expects a 5D tensor with shape: (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}")
                    InputShape = InputShape[-1:] + InputShape[:-1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temph = -(InputShape[1] // -strides_height) #To make it work as ceiling
                    tempw = -(InputShape[2] // -strides_width)
                    tempd = -(InputShape[3] // -strides_depth)
                    return (HomaTect_layers_list[layerid]["filters"], temph, tempw, tempd)
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = ((InputShape[1] - kernel_height) // strides_height) + 1
                    tempw = ((InputShape[2] - kernel_width) // strides_width) + 1
                    tempd = ((InputShape[3] - kernel_depth) // strides_depth) + 1
                    return (HomaTect_layers_list[layerid]["filters"], temph, tempw, tempd)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True:
                # ************************* if len(InputShape) != 4: # Check input shape of the Conv3D layer
                    print(f"Layer{layerid+1}: A Conv3D layer with channels_last format, expects a 5D tensor with shape: (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temph = -(InputShape[0] // -strides_height) #To make it work as ceiling
                    tempw = -(InputShape[1] // -strides_width)
                    tempd = -(InputShape[2] // -strides_depth)
                    return (temph, tempw, tempd, HomaTect_layers_list[layerid]["filters"])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = ((InputShape[0] - kernel_height) // strides_height) + 1
                    tempw = ((InputShape[1] - kernel_width) // strides_width) + 1
                    tempd = ((InputShape[2] - kernel_depth) // strides_depth) + 1
                    return (temph, tempw, tempd, HomaTect_layers_list[layerid]["filters"])
        
    if HomaTect_layers_list[layerid]["layer"] in ["MaxPooling1D", "MaxPooling2D", "MaxPooling3D", "AveragePooling1D", "AveragePooling2D", "AveragePooling3D"]:
        if HomaTect_layers_list[layerid]["layer"] in ["MaxPooling1D", "AveragePooling1D"]:
            # pool_size: int.
            # strides: int or None. If None, it will default to pool_size.
            # padding: "valid" or "same"
            # data_format: "channels_last": (batch, steps, features) while "channels_first": (batch, features, steps). If you never set it, then it will be "channels_last".
            # Input shape: data_format="channels_last": 3D tensor with shape (batch_size, steps, features).
                         # data_format="channels_first": 3D tensor with shape (batch_size, features, steps).
            # Output shape: data_format="channels_last": 3D tensor with shape (batch_size, downsampled_steps, features).
                        # data_format="channels_first": 3D tensor with shape (batch_size, features, downsampled_steps). 
            
            if isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides = HomaTect_layers_list[layerid]["strides"]                  
            elif HomaTect_layers_list[layerid]["strides"] is None:
                strides = HomaTect_layers_list[layerid]["pool_size"]

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # ******************** if len(InputShape) != 2: # Check input shape of the MaxPooling1D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_first format, expects a 3D tensor with shape: (batch_size, channels, steps) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}")
                    InputShape = InputShape[-1:] + InputShape[:-1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temp = -(InputShape[1] // -strides ) # To make it work as ceiling
                    return (InputShape[0], temp)
                elif HomaTect_layers_list[layerid]["padding"] == "valid": # output_shape = ceil(input_shape - pool_size + 1) / strides)
                    temp = -((InputShape[1] - HomaTect_layers_list[layerid]["pool_size"] + 1) // -strides)
                    return (InputShape[0], temp)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True:
                # ******************* if len(InputShape) != 2: # Check input shape of the MaxPooling1D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_last format, expects a 3D tensor with shape: (batch_size, steps, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same":
                    temp = -(InputShape[0] // -strides ) # To make it work as ceiling
                    return (temp, InputShape[1])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temp = -((InputShape[0] - HomaTect_layers_list[layerid]["pool_size"] + 1) // -strides)
                    return (temp, InputShape[1])

        if HomaTect_layers_list[layerid]["layer"] in ["MaxPooling2D", "AveragePooling2D"]:
            # pool_size:  int or tuple of 2 integers.
            # strides:  int or tuple of 2 integers, or None. If None, it will default to pool_size.
            # padding: "valid" or "same"
            # data_format: "channels_last": (batch, height, width, channels) while "channels_first": (batch, channels, height, width). If you never set it, then it will be "channels_last".
            # Input shape: data_format="channels_last": 4D tensor with shape (batch_size, height, width, channels).
                         # data_format="channels_first": 4D tensor with shape (batch_size, channels, height, width).
            # Output shape: data_format="channels_last": 4D tensor with shape (batch_size, downsampled_steps, features).
                        # data_format="channels_first": 4D tensor with shape (batch_size, features, downsampled_steps). 
            
            if isinstance(HomaTect_layers_list[layerid]["pool_size"], tuple):
                pool_height = HomaTect_layers_list[layerid]["pool_size"][0]
                pool_width = HomaTect_layers_list[layerid]["pool_size"][1]
            elif isinstance(HomaTect_layers_list[layerid]["pool_size"], int):
                pool_height = HomaTect_layers_list[layerid]["pool_size"]
                pool_width = HomaTect_layers_list[layerid]["pool_size"]
                
            if isinstance(HomaTect_layers_list[layerid]["strides"], tuple):
                strides_height = HomaTect_layers_list[layerid]["strides"][0]
                strides_width = HomaTect_layers_list[layerid]["strides"][1]
            if isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides_height = HomaTect_layers_list[layerid]["strides"]
                strides_width = HomaTect_layers_list[layerid]["strides"]

            if HomaTect_layers_list[layerid]["strides"] is None:
                strides_height = pool_height
                strides_width = pool_width

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # *************** if len(InputShape) != 3: # Check input shape of the MaxPooling2D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_first format, expects a 4D tensor with shape: (batch_size, channels, height, width) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}")
                    InputShape = InputShape[-1:] + InputShape[:-1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temph = -(InputShape[1] // -strides_height) # To make it work as ceiling
                    tempw = -(InputShape[2] // -strides_width)
                    return (InputShape[0], temph, tempw)
                elif HomaTect_layers_list[layerid]["padding"] == "valid": # output_shape = ceil(input_shape - pool_size + 1) / strides)
                    temph = -((InputShape[1] - pool_height + 1) // -strides_height)
                    tempw = -((InputShape[2] - pool_width + 1) // -strides_width)
                    return (InputShape[0], temph, tempw)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True:
                # ********************** if len(InputShape) != 3: # Check input shape of the MaxPooling2D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_last format, expects a 4D tensor with shape: (batch_size, height, width, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same":
                    temph = -(InputShape[0] // -strides_height) # To make it work as ceiling
                    tempw = -(InputShape[1] // -strides_width)
                    return (temph, tempw, InputShape[2])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = -((InputShape[0] - pool_height + 1) // -strides_height)
                    tempw = -((InputShape[1] - pool_width + 1) // -strides_width)
                    return (temph, tempw, InputShape[2])


        if HomaTect_layers_list[layerid]["layer"] in ["MaxPooling3D", "AveragePooling3D"]:
            # pool_size:  int or tuple of 3 integers.
            # strides:  int or tuple of 3 integers, or None. If None, it will default to pool_size.
            # padding: "valid" or "same"
            # data_format: "channels_last": (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels) while "channels_first": (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3). If you never set it, then it will be "channels_last".
            # Input shape: data_format="channels_last": 5D tensor with shape (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels)
                         # data_format="channels_first": 5D tensor with shape (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3)
            # Output shape: data_format="channels_last": 5D tensor with shape (batch_size, pooled_dim1, pooled_dim2, pooled_dim3, channels)
                        # data_format="channels_first": 5D tensor with shape (batch_size, channels, pooled_dim1, pooled_dim2, pooled_dim3) 
            
            if isinstance(HomaTect_layers_list[layerid]["pool_size"], tuple):
                pool_height = HomaTect_layers_list[layerid]["pool_size"][0]
                pool_width = HomaTect_layers_list[layerid]["pool_size"][1]
                pool_depth = HomaTect_layers_list[layerid]["pool_size"][2]
            elif isinstance(HomaTect_layers_list[layerid]["pool_size"], int):
                pool_height = HomaTect_layers_list[layerid]["pool_size"]
                pool_width = HomaTect_layers_list[layerid]["pool_size"]
                pool_depth = HomaTect_layers_list[layerid]["pool_size"]
                
            if isinstance(HomaTect_layers_list[layerid]["strides"], tuple):
                strides_height = HomaTect_layers_list[layerid]["strides"][0]
                strides_width = HomaTect_layers_list[layerid]["strides"][1]
                strides_depth = HomaTect_layers_list[layerid]["strides"][2]
            if isinstance(HomaTect_layers_list[layerid]["strides"], int):
                strides_height = HomaTect_layers_list[layerid]["strides"]
                strides_width = HomaTect_layers_list[layerid]["strides"]
                strides_depth = HomaTect_layers_list[layerid]["strides"]

            if HomaTect_layers_list[layerid]["strides"] is None:
                strides_height = pool_height
                strides_width = pool_width
                strides_depth = pool_width

            if HomaTect_layers_list[layerid]["data_format"] == "channels_first":
                if user_input_shape != InputShape[1:] and HomaTect_first_conv == True:
                # ***************** if len(InputShape) != 4: # Check input shape of the MaxPooling3D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_first format, expects a 5D tensor with shape: (batch_size, channels, spatial_dim1, spatial_dim2, spatial_dim3) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[-1:] + InputShape[:-1]}")
                    InputShape = InputShape[-1:] + InputShape[:-1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same": # output_spatial_shape[i] = ceil(input_spatial_shape[i] / strides[i])
                    temph = -(InputShape[1] // -strides_height) # To make it work as ceiling
                    tempw = -(InputShape[2] // -strides_width)
                    tempd = -(InputShape[3] // -strides_depth)
                    return (InputShape[0], temph, tempw, tempd)
                elif HomaTect_layers_list[layerid]["padding"] == "valid": # output_shape = ceil(input_shape - pool_size + 1) / strides)
                    temph = -((InputShape[1] - pool_height + 1) // -strides_height)
                    tempw = -((InputShape[2] - pool_width + 1) // -strides_width)
                    tempd = -((InputShape[3] - pool_depth + 1) // -strides_depth)
                    return (InputShape[0], temph, tempw, tempd)

            else: # If not specified then channels_last
                if user_input_shape != InputShape[:-1] and HomaTect_first_conv == True:
                # ******************* if len(InputShape) != 4: # Check input shape of the MaxPooling2D layer without batch_size
                    print(f"Layer{layerid+1}: A {HomaTect_layers_list[layerid]['layer']} layer with channels_last format, expects a 5D tensor with shape: (batch_size, spatial_dim1, spatial_dim2, spatial_dim3, channels) while it receivedan input shape as {(None,) + InputShape}. Use the correct channels formatting or input_shape. Debugging continues with the fix applied as {(None,) + InputShape[1:] + InputShape[:1]}")
                    InputShape = InputShape[1:] + InputShape[:1]
                HomaTect_first_conv = False
                if HomaTect_layers_list[layerid]["padding"] == "same":
                    temph = -(InputShape[0] // -strides_height) # To make it work as ceiling
                    tempw = -(InputShape[1] // -strides_width)
                    tempd = -(InputShape[2] // -strides_depth)
                    return (temph, tempw, tempd, InputShape[3])
                elif HomaTect_layers_list[layerid]["padding"] == "valid":
                    temph = -((InputShape[0] - pool_height + 1) // -strides_height)
                    tempw = -((InputShape[1] - pool_width + 1) // -strides_width)
                    tempd = -((InputShape[2] - pool_depth + 1) // -strides_depth)
                    return (temph, tempw, tempd, InputShape[3])
        
    if HomaTect_layers_list[layerid]["layer"] == "Flatten":
        # If inputs are shaped (batch,) without a feature axis, then flattening adds an extra channel dimension and output shape is (batch, 1).
        temp = 1
        neg = False
        if len(InputShape) == 0:
            return (1,)
        else:
            for element in InputShape:
                if element < 0:
                    neg = True
                temp = temp * element
            if neg == True and temp >= 0: # so that even number of negative dimensions dont make it positive
                temp = -temp
            return (temp,)
            
    if HomaTect_layers_list[layerid]["layer"] == "Reshape":
        # target_shape: Target shape. Tuple of integers, does not include the samples dimension (batch size).
        temp1 = 1
        temp2 = 1
        infered = 0
        target = ()
        neg = False
        for element in InputShape:
            if element < 0:
                neg = True
            temp1 = temp1 * element
        if neg == True and temp1 >= 0: # so that more than one negative dimension dont make it positive
            temp1 = -temp1

        for element in HomaTect_layers_list[layerid]["target_shape"]:
            temp2 = temp2 * element
        temp2 = abs(temp2) # becasue it may contain -1
        
        for element in HomaTect_layers_list[layerid]["target_shape"]:
            if element != -1:
                target = target + (element,)
            else:
                if infered != 0:
                     raise ValueError(f" layer{layerid+1}: More than one dimention is set to -1, while only one dimension can be inferred")
                if temp1 % temp2 != 0:
                    raise ValueError(f"Your input shape is {InputShape} and you want to reshape it to {HomaTect_layers_list[layerid]['target_shape']}. The product of input tensor dimensions should be equal to the product of target tensor dimensions")
                infered = temp1 // temp2
                target = target + (infered,)
                resolved_shape = []
                for dim in HomaTect_layers_list[layerid]["target_shape"]:
                    if dim == -1:
                        resolved_shape.append(infered)
                    else:
                        resolved_shape.append(dim)
                temp2 = np.prod(resolved_shape)

        if temp1 != temp2:
            raise ValueError(f"Your input shape is {InputShape} and you want to reshape it to {HomaTect_layers_list[layerid]['target_shape']}. The product of input tensor dimensions should be equal to the product of target tensor dimensions")
        if temp1 < 0:
            print(f"Layer{layerid+1}: Cant perform reshape on negative input dimensions so returning {InputShape}")
            return InputShape # cant perform reshape on negative dimensions
        else:
            return target

    if HomaTect_layers_list[layerid]["layer"] in ["Dropout", "Activation", "BatchNormalization", "elu", "leakyrelu", "prelu", "relu", "softmax"]:
        return InputShape
      
```
